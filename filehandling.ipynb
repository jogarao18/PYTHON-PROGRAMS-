```python
#file handling
file=open("new.txt","w+")
file.write("this is my new file")
file.seek(0)
data=file.read();
print(data)
file.close()
```

```python
with open("new.txt","a+") as f:
  f.write("\n")
  f.write("python is a complex subject")
  f.seek(0)
  data=f.read()
print(data)
```

```python
#example
file=open("f","w+")
file.write("this is my new file")
file.seek(0)
data=file.read()
print(data)
file.write("\npython 🐍")
file.seek(0)
data=file.read()
print(data)
```

```python
with open("f","r") as F:
  print(F.readline())
  print(F.readline())
```
