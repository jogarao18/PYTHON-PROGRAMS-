```python
#list operations
#concetation
l1=[1,2,4,3,77,"mani"]
print(l1)
l2=[55,20,12,88,"sai"]
print(l2)
#concetation
print(l1+l2)
#repetation
print(l1*2)
#indexing
print(l1[1])
print(l1[5])
print(l1[-1])
#slicing
print(l1[:])
print(l1[0:6])
print(l1[0:])
print(l1[:6])
#slicing for reverse
print(l1[::-1])
#membership operator
if 2 not in l2:
  print("yeahh")
else:
  print("noo")
if 20 in l2:
  print("yeahh")
else:
  print("noo")
#iteration
for i in l2:
  print(i,end=" ")
print(" ")
for i in range(len(l2)):
  print(l2[i],end=" ")
```

```python
#built in functions
l1=[10,77,5,40,50]
print(len(l1))
print(sum(l1))
print(max(l1))
print(min(l1))
print(sorted(l1))
print(list(reversed(l1)))
print(l1)
```

```python
#list methods
l1=[10,5,2,99,88]
l2=[100,11,87,93]
l1.append(100)
print(l1)
l1.remove(100)
print(l1)
l1.insert(5,77)
print(l1)
l1.pop()
print(l1)
l1.pop(3)
print(l1)
l1.sort()
print(l1)
print(l1.count(2))
print(l1.index(5))
l1.extend(l2)
print(l1)
```

```python
#sum of list
l1=[1000,13,55,267,49]
n=len(l1)
sum=0
for i in range(n):
  sum=sum+l1[i]
print(sum)
```

```python
#print even elements
for i in range(n):
  if l1[i]%2==0:
    print(l1[i])
```

```python
#print odd elements
for i in range(n):
  if l1[i]%2!=0:
    print(l1[i],end=" ")
```
