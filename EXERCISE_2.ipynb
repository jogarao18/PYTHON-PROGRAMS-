```python
# MATCH CASE
x=int(input("enter the day:"))
match x:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid")
```

```python
#sum of digits
x=int(input("enter num1 :"))
sum=0
r=0
while(x>0):
  r=x%10
  sum=sum+r
  x=x//10
print(f"sum of digits is {sum}")
```

```python
#reverse of number
x=int(input("enter num1 :"))
res=0
r=0
while(x>0):
  r=x%10
  res=res*10+r
  x=x//10
print(f"reverse of nums is {res}")
```

```python
#prime or not;
x=int(input("enter num :"))
sum=0
for i in range(1,x+1,1):
    if(x%i==0):
      sum=sum+1
    elif(sum>2):
      break
if(x==0):
  print("either prime or compisite")
elif(sum==2):
  print(f"{x} is a prime")
else:
  print(f"{x} is not a prime")
```
