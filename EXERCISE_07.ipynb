```python
#SETS OPERATIONS
a={1,2,3}
b={3,4,5}
print(a|b)
print(a&b)
print(a-b)
print(a^b)
#can be performed with or without two sets
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.union([7,8]))
print(a.intersection([4,5]))
```

```python
a=set(int(i) for i in input().split())
b=set(int(i) for i in input().split())
a.add(100) #addd
print(a)
a.update([1000,2000]) #update
print(a)
a.remove(100)
print(a)
a.discard(1000)
print(a)
print("Union:", a.union(b))
print("Intersection:", a.intersection(b))
print("A - B:", a.difference(b))
print("Symmetric Difference:", a.symmetric_difference(b))
```
