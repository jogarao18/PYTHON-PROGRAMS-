```python
#write a python progarm to count no of duplicates
l=[int (i) for i in input().split()]
s=set(l)
print(len(l)-len(s))
```

```python
#write a python program to find no of numbers having duplicates
l=[int (i) for i in input().split()]
l2=[]
for i in l:
  if(l.count(i)>1):
    l2.append(i)
print(len(set(l2)))
```

```python
#to count the number of unique elements
l=[int (i) for i in input().split()]
l2=[]
for i in l:
  if(l.count(i)==1):
    l2.append(i)
print(len(set(l2)))
```
