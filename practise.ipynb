```python
#dictionarys
d={"sno":1,"name":"mani","age":18}
print(d)
print(d["sno"])
d["marks"]=50
print(d)
print(d.get("name"))
d.pop("marks")
print(d)
print(list(d.keys()))
print(list(d.values()))
print(list(d.items()))
d.update({"name":"sai","age":21})
print(d)
```

```python
k=[i for i in input().split()]
v=[int(i) for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)
```

```python
#write a program frequency and occurance element of each list
l=[1,2,3,5,1,4,6,2,1,1,3,2,5,4]
d={}
for i in l:
  if i in d:
    d[i]=d[i]+1;
  else:
    d[i]=1
print(d)
```
